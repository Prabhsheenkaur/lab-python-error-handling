````python

def initialize_inventory(products):
    """
    Prompts user for quantities of products.
    Ensures non-negative integer input.
    Uses dict comprehension.
    """
    def get_valid_quantity(product):
        while True:
            try:
                qty = int(input(f"Enter the quantity of {product}s available: "))
                if qty < 0:
                    print("Quantity cannot be negative. Try again.")
                    continue
                return qty
            except ValueError:
                print("Invalid input. Please enter a whole number (e.g., 0, 5, 12).")

    inventory = {product: get_valid_quantity(product) for product in products}
    return inventory


def get_customer_orders(inventory):
    """
    Prompts user for number of orders and product names.
    Ensures:
    - valid number of orders (non-negative int)
    - product exists in inventory
    - product is in stock (>0)
    Returns a set of ordered product names (1 unit each).
    """
    while True:
        try:
            n = int(input("How many different products does the customer want to order? "))
            if n < 0:
                print("Number of orders cannot be negative. Try again.")
                continue
            break
        except ValueError:
            print("Invalid input. Please enter a whole number (e.g., 1, 2, 3).")

    orders = set()
    available_products = set(inventory.keys())

    for i in range(1, n + 1):
        while True:
            product = input(f"Enter product #{i} name: ").strip()

            if product not in available_products:
                print("That product is not in the inventory list. Try again.")
                print("Available:", ", ".join(sorted(available_products)))
                continue

            if inventory.get(product, 0) <= 0:
                print(f"'{product}' is currently out of stock. Choose another product.")
                continue

            if product in orders:
                print("You already added that product (assume max 1 unit each). Choose another.")
                continue

            orders.add(product)
            break

    return orders


def calculate_total_price(customer_orders):
    """
    Prompts user for the price of each ordered product.
    Ensures:
    - valid float
    - non-negative
    Uses comprehension to compute total.
    """
    def get_valid_price(product):
        while True:
            try:
                price = float(input(f"Enter the price of '{product}': "))
                if price < 0:
                    print("Price cannot be negative. Try again.")
                    continue
                return price
            except ValueError:
                print("Invalid input. Please enter a number (e.g., 2.50, 10, 99.99).")

    prices = {product: get_valid_price(product) for product in customer_orders}
    total = sum(prices.values())
    return total, prices


def update_inventory(inventory, customer_orders):
    """
    Decrease inventory by 1 for each ordered product.
    Remove products with 0 quantity.
    Uses dict comprehension to filter remaining inventory.
    """
    for product in customer_orders:
        inventory[product] -= 1

   
    inventory = {p: q for p, q in inventory.items() if q > 0}
    return inventory


def main():
    products = ["apple", "banana", "orange", "pear", "mango"]

    print("\n--- Initialize Inventory ---")
    inventory = initialize_inventory(products)

    print("\nCurrent inventory:", inventory)

    print("\n--- Customer Orders ---")
    customer_orders = get_customer_orders(inventory)
    print("Customer ordered:", customer_orders)

    print("\n--- Total Price ---")
    total_price, prices = calculate_total_price(customer_orders)
    print("Prices:", prices)
    print("Total price:", total_price)

    print("\n--- Update Inventory ---")
    inventory = update_inventory(inventory, customer_orders)
    print("Updated inventory:", inventory)


if __name__ == "__main__":
    main()
````
